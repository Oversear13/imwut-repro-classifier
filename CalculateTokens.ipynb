```python
import os
import tiktoken
from pypdf import PdfReader
import pandas as pd
from tqdm import tqdm

def analyze_bachelor_studies(base_path, model="gpt-4o"):
    all_results = []
    encoding = tiktoken.encoding_for_model(model)
    
    # Die spezifischen Unterordner (Jahre)
    years = ["2023", "2024", "2025"]
    
    print(f"Starte Analyse in: {base_path}\n")

    for year in years:
        year_path = os.path.join(base_path, year)
        year_tokens = 0
        
        if not os.path.exists(year_path):
            print(f"Warnung: Ordner für {year} nicht gefunden unter {year_path}")
            continue
            
        # PDF-Dateien im Jahresordner finden
        pdf_files = [f for f in os.listdir(year_path) if f.endswith('.pdf')]
        
        for file_name in tqdm(pdf_files, desc=f"Verarbeite Jahr {year}"):
            full_path = os.path.join(year_path, file_name)
            
            try:
                reader = PdfReader(full_path)
                file_token_count = 0
                
                for page in reader.pages:
                    page_text = page.extract_text()
                    if page_text:
                        file_token_count += len(encoding.encode(page_text))
                
                year_tokens += file_token_count
                
                all_results.append({
                    "Jahr": year,
                    "Dateiname": file_name,
                    "Seiten": len(reader.pages),
                    "Tokens": file_token_count
                })
                
            except Exception as e:
                print(f" Fehler bei {file_name}: {e}")
        
        # Zwischenergebnis für das Jahr drucken
        print(f"-> Jahr {year} abgeschlossen: {year_tokens:,} Tokens".replace(',', '.'))
        print("-" * 30)

    return pd.DataFrame(all_results)

# --- Pfad Anpassung ---
base_path = "/home/Mike-bazz/Dokumente/Meine_Bachelorarbeit/Studien/Studien/"

# Analyse ausführen
df_studien = analyze_bachelor_studies(base_path)

# --- Finale Auswertung ---
if not df_studien.empty:
    total_tokens = df_studien['Tokens'].sum()
    
    print("\n" + "="*40)
    print("FINALE AUSWERTUNG")
    print("="*40)
    
    # Gruppierte Summen pro Jahr anzeigen
    summary_per_year = df_studien.groupby('Jahr')['Tokens'].sum()
    for year, tokens in summary_per_year.items():
        print(f"Summe Tokens {year}: {tokens:>12,}".replace(',', '.'))
    
    print("-" * 40)
    print(f"GESAMTANZAHL TOKENS: {total_tokens:>12,}".replace(',', '.'))
    print("="*40)
    
    # Tabelle anzeigen
    display(df_studien)
else:
    print("Keine Daten gefunden.")
```

```python
import os
import tiktoken
from pypdf import PdfReader
import pandas as pd
from tqdm import tqdm
import warnings
import re

warnings.filterwarnings("ignore")

def analyze_bachelor_studies_final(base_path, embedding_model="text-embedding-3-large"):
    all_results = []
    
    # Der korrekte Tokenizer für die v3-Embedding Modelle
    encoding = tiktoken.encoding_for_model(embedding_model)
    
    pdf_tasks = []
    for root, _, files in os.walk(base_path):
        for file in files:
            if file.lower().endswith(".pdf"):
                pdf_tasks.append((root, file))
    
    print(f"Gefundene Dateien insgesamt: {len(pdf_tasks)}")

    for root, file_name in tqdm(pdf_tasks, desc="Verarbeite Dateien"):
        full_path = os.path.join(root, file_name)
        folder_year = os.path.basename(root)
        
        # Extrahiere die Nummer aus dem Dateinamen (z.B. 146 aus "2023_Dec_146.pdf")
        # Wir suchen nach einer Ziffernfolge am Ende vor dem .pdf
        match = re.search(r'(\d+)\.pdf$', file_name, re.IGNORECASE)
        file_number = int(match.group(1)) if match else 0
        
        try:
            reader = PdfReader(full_path, strict=False)
            file_token_count = 0
            
            for page in reader.pages:
                try:
                    text = page.extract_text()
                    if text:
                        file_token_count += len(encoding.encode(text))
                except:
                    continue 
            
            all_results.append({
                "Jahr": folder_year,
                "Nr": file_number,  # Für die Sortierung
                "Dateiname": file_name,
                "Seiten": len(reader.pages),
                "Tokens": file_token_count
            })
        except:
            continue

    df = pd.DataFrame(all_results)
    
    # Sortierung: Erst nach Jahr, dann nach der extrahierten Nummer
    if not df.empty:
        df = df.sort_values(by=['Jahr', 'Nr']).reset_index(drop=True)
        
    return df

# --- Ausführung ---
path = "/home/Mike-bazz/Dokumente/Meine_Bachelorarbeit/Studien/Studien/"
df_studien = analyze_bachelor_studies_final(path)

# --- Auswertung ---
if not df_studien.empty:
    print("\n" + "="*45)
    print(f"TOKEN-AUSWERTUNG (Modell: text-embedding-3-large)")
    print("="*45)
    
    summary = df_studien.groupby('Jahr')['Tokens'].sum()
    for year in sorted(summary.index):
        print(f"Jahr {year}: {summary[year]:>20,}".replace(',', '.'))
    
    print("-" * 45)
    print(f"GESAMT-TOKENS: {df_studien['Tokens'].sum():>18,}".replace(',', '.'))
    print("="*45)
    
    pd.set_option('display.max_rows', None)
    display(df_studien)
```
